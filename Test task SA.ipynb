## Схема доставки Push-уведомления:

***Событийно-ориентированная архитектура (EDA)*** — лучшая для пуш-уведомлений, так как обеспечивает высокую скорость отправки, асинхронность и масштабируемость. EDA разделяет системы: сервис-источник (например, заказ) генерирует событие, а сервис уведомлений обрабатывает его, не задерживая основной процесс. Это гарантирует надежность при пиковых нагрузках.




##### Как это работает:
* ***Событие***: В «Сервисе Заказов» статус заказа меняется на «Отменен».
* ***Очередь***: Сервис кидает сообщение с айди заказа, статусом и юзер-айди в Kafka.
* ***Обработка***: «Сервис уведомлений» забирает это сообщение.
* ***Данные***: Сервис уведомлений запрашивает в базе данных актуальные push_token для нужного юзер-айди.
* ***Вендор***: Если у пользователя Android, сервис уведомлений делает POST-запрос к API Firebase.
* ***Доставка***: Firebase доставляет пуш на телефон пользователя.
